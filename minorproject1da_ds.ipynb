```python
import numpy as np
from datetime import datetime, timedelta

print("=" * 65)
print(" GROUP DNA WHATSAPP ANALYZER")
print("=" * 65)
```

```python
FILE_NAME = "/content/whatsappchat.csv"

try:
    file = open(FILE_NAME, "r", encoding="utf-8-sig")
except FileNotFoundError:
    try:
        file = open("/content/whatsappchatcsv", "r", encoding="utf-8-sig")
        FILE_NAME = "/content/whatsappchatcsv"
    except FileNotFoundError:
        file = open("/content/mywhatsappchat.csv", "r", encoding="utf-8-sig")
        FILE_NAME = "/content/mywhatsappchat.csv"

lines = file.readlines()
file.close()

print("File loaded successfully!")
print("File:", FILE_NAME)
print("Total lines:", len(lines))
```

```python
messages = []

for line in lines[1:]:
    line = line.strip()

    if line == "":
        continue

    parts = line.split(",", 3)

    if len(parts) == 4:
        date = parts[0].strip()
        time = parts[1].strip()
        sender = parts[2].strip()
        text = parts[3].strip()

        messages.append({
            "date": date,
            "time": time,
            "sender": sender,
            "text": text
        })

print("Messages parsed:", len(messages))
```

```python
valid_messages = []

for msg in messages:

    try:
        msg["timestamp"] = datetime.strptime(
            msg["date"] + " " + msg["time"],
            "%d/%m/%Y %H:%M"
        )

        valid_messages.append(msg)

    except ValueError:
        pass

messages = valid_messages

print("Valid messages:", len(messages))

for msg in messages[:5]:
    print(
        msg["date"],
        msg["time"],
        "-",
        msg["sender"],
        ":",
        msg["text"]
    )
```

```python
participants = []
message_count = {}

for msg in messages:

    person = msg["sender"]

    if person not in participants:
        participants.append(person)

    message_count[person] = message_count.get(person, 0) + 1

participants = sorted(participants)

ranking = sorted(
    message_count.items(),
    key=lambda x: x[1],
    reverse=True
)

print("=" * 65)
print("PARTICIPANTS:", len(participants))
print("=" * 65)

for person, count in ranking:
    print(f"{person:<15} : {count} messages")
```

```python
date_count = {}

for msg in messages:

    day = msg["timestamp"].date()

    date_count[day] = date_count.get(day, 0) + 1

if date_count:

    first_date = min(date_count)
    last_date = max(date_count)

    busiest_day = max(
        date_count,
        key=date_count.get
    )

    print("=" * 65)
    print("DATE ANALYSIS")
    print("=" * 65)

    print("First Date :", first_date)
    print("Last Date  :", last_date)
    print(
        "Total Days :",
        (last_date - first_date).days + 1
    )
    print(
        "Busiest Day:",
        busiest_day
    )
    print(
        "Messages   :",
        date_count[busiest_day]
    )

else:
    print("No valid date data found.")
```

```python
hour_count = {}

for msg in messages:

    hour = msg["timestamp"].hour

    hour_count[hour] = hour_count.get(hour, 0) + 1

if hour_count:

    busiest_hour = max(
        hour_count,
        key=hour_count.get
    )

    print("=" * 65)
    print("HOURLY ACTIVITY")
    print("=" * 65)

    for hour in sorted(hour_count):
        print(
            f"{hour:02d}:00 - "
            f"{hour_count[hour]} messages"
        )

    print(
        "\nBusiest Hour:",
        f"{busiest_hour:02d}:00"
    )
```

```python
people = sorted(participants)

person_index = {}

for i in range(len(people)):
    person_index[people[i]] = i

activity = np.zeros(
    (len(people), 24),
    dtype=int
)

for msg in messages:

    row = person_index[msg["sender"]]
    hour = msg["timestamp"].hour

    activity[row][hour] += 1

print("=" * 65)
print("NUMPY ACTIVITY MATRIX")
print("=" * 65)

print("Shape:", activity.shape)
print(activity)
```

```python
print("=" * 65)
print("                    ACTIVITY HEATMAP")
print("=" * 65)

print("             ", end="")

for hour in range(24):
    print(f"{hour:02d} ", end="")

print()

for i in range(len(people)):

    print(f"{people[i]:<12}", end="")

    row_max = np.max(activity[i])

    for value in activity[i]:

        if value == 0:
            symbol = "."
        elif row_max > 0 and value <= row_max * 0.25:
            symbol = "░"
        elif row_max > 0 and value <= row_max * 0.50:
            symbol = "▒"
        else:
            symbol = "█"

        print(f"{symbol}  ", end="")

    print()
```

```python
stop_words = {
    "the", "is", "a", "an", "and", "or",
    "to", "of", "in", "on", "for", "it",
    "this", "that", "you", "we", "are",
    "was", "be", "am", "i", "me", "my"
}

word_count = {}

for msg in messages:

    words = msg["text"].lower().split()

    for word in words:

        word = word.strip(".,!?;:'\"()[]{}")

        if word == "":
            continue

        if word in stop_words:
            continue

        word_count[word] = word_count.get(word, 0) + 1

top_words = sorted(
    word_count.items(),
    key=lambda x: x[1],
    reverse=True
)[:10]

print("=" * 65)
print("TOP 10 WORDS")
print("=" * 65)

for word, count in top_words:
    print(f"{word:<20} : {count}")
```

```python
response_times = {}

for person in participants:
    response_times[person] = []

previous = None

for msg in messages:

    if previous is not None:

        if msg["sender"] != previous["sender"]:

            gap = (
                msg["timestamp"]
                - previous["timestamp"]
            ).total_seconds()

            if gap >= 0 and gap <= 86400:

                response_times[
                    msg["sender"]
                ].append(gap)

    previous = msg

average_response = {}

for person in participants:

    gaps = response_times[person]

    if len(gaps) > 0:
        average_response[person] = (
            sum(gaps) / len(gaps)
        )
    else:
        average_response[person] = 0

print("=" * 65)
print("RESPONSE SPEED")
print("=" * 65)

for person in participants:

    minutes = average_response[person] / 60

    print(
        f"{person:<15} : "
        f"{minutes:.1f} minutes"
    )
```

```python
all_days = []

if date_count:

    current = first_date

    while current <= last_date:

        all_days.append(current)

        current += timedelta(days=1)

active_days = {}

for person in participants:
    active_days[person] = set()

for msg in messages:

    active_days[
        msg["sender"]
    ].add(
        msg["timestamp"].date()
    )

silent_streak = {}

for person in participants:

    current_streak = 0
    maximum_streak = 0

    for day in all_days:

        if day in active_days[person]:

            current_streak = 0

        else:

            current_streak += 1

            if current_streak > maximum_streak:
                maximum_streak = current_streak

    silent_streak[person] = maximum_streak

print("=" * 65)
print("THE GHOST — LONGEST SILENT STREAK")
print("=" * 65)

for person, days in sorted(
    silent_streak.items(),
    key=lambda x: x[1],
    reverse=True
):

    print(
        f"{person:<15} : "
        f"{days} silent days"
    )
```

```python
archetypes = {}

funny_words = [
    "haha", "hahaha", "lol",
    "lmao", "rofl", "lmfao"
]

for person in participants:

    entertainer = 0
    drama = 0
    night = 0
    storyteller_words = 0

    person_msgs = []

    for msg in messages:

        if msg["sender"] == person:

            person_msgs.append(msg)

            text = msg["text"]
            lower = text.lower()

            storyteller_words += len(text.split())

            for word in funny_words:

                if word in lower:
                    entertainer += 1

            if len(text) >= 3 and text.isupper():
                drama += 1

            if text.count("!") >= 2:
                drama += 1

            hour = msg["timestamp"].hour

            if hour >= 23 or hour <= 4:
                night += 1

    if len(person_msgs) > 0:
        storyteller = (
            storyteller_words /
            len(person_msgs)
        )
    else:
        storyteller = 0

    scores = {
        "THE GHOST": silent_streak.get(person, 0),
        "THE ENTERTAINER": entertainer,
        "THE DRAMA QUEEN": drama,
        "THE NIGHT OWL": night,
        "THE STORYTELLER": storyteller
    }

    archetypes[person] = max(
        scores,
        key=scores.get
    )

print("=" * 65)
print("PERSONALITY ARCHETYPES")
print("=" * 65)

for person in participants:

    print(
        f"{person:<15} → "
        f"{archetypes[person]}"
    )
```
